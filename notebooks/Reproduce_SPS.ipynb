# Reproduce the Sequence Persistence Score

This notebook walks through the evaluation used in the PulseAI working paper:

> Ananya Das. (2026). *Identity Persistence in AI-Generated Video: A Sequence Evaluation Framework for Founder-Led Outbound*. PulseAI Working Paper. Zenodo.

You need a consented source portrait and 3–5 generated videos to run the full evaluation.

```python
import sys
sys.path.append('..')

from pathlib import Path
from eval.extract_faces import load_model, get_face_embedding, get_face_embedding_from_video
from eval.extract_style import get_style_vector, get_style_vector_from_video
from eval.compute_sps import compute_sps
```

## 1. Load the source portrait and generated videos

```python
source_image = Path('../data/sample/source.jpg')
videos = [
    Path('../data/sample/touch1.mp4'),
    Path('../data/sample/touch2.mp4'),
    Path('../data/sample/touch3.mp4'),
]

print(f"Source: {source_image}")
print(f"Videos: {videos}")
```

## 2. Extract face embeddings and style features

```python
model = load_model()

source_face = get_face_embedding(source_image, model)
source_style = get_style_vector(source_image)

face_embeddings = [get_face_embedding_from_video(v, model) for v in videos]
style_vectors = [get_style_vector_from_video(v) for v in videos]

print(f"Source face embedding shape: {source_face.shape}")
print(f"Source style vector shape: {source_style.shape}")
```

## 3. Compute the Sequence Persistence Score

```python
result = compute_sps(source_face, face_embeddings, source_style, style_vectors)
result
```

## Interpretation

- **Face persistence > 0.65**: the sender's face is recognisably the same across the sequence.
- **SPS > 0.70**: the sequence as a whole passes the identity-consistency check.

If either threshold fails, the sequence should be regenerated or reviewed.
